# Classroom Attendance Prediction Using Academic Schedule and Historical Attendance Data


## Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, LabelEncoder, FunctionTransformer

from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier, GradientBoostingRegressor
from sklearn.svm import SVR, SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from xgboost import XGBRegressor, XGBClassifier

from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score, mean_absolute_percentage_error,
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    confusion_matrix, ConfusionMatrixDisplay, roc_curve
)

pd.set_option("display.width", 140)
plt.rcParams["figure.figsize"] = (8, 4)


def to_dense_array(x):
    """Convert a sparse matrix (from OneHotEncoder) to a dense array.
    Needed because GaussianNB cannot work with sparse input. This is a
    named function, not a lambda, so the trained pipeline can be saved
    with joblib and re-loaded later in the Streamlit app."""
    return x.toarray() if hasattr(x, "toarray") else x
```

## Step 2: Load Dataset

```python
df = pd.read_csv("attendense.csv")
print("Dataset shape:", df.shape)
df.head()
```

## Step 3: Understand the Dataset

We check the shape, column names, data types, missing values, duplicate rows, and
basic statistics before changing anything.


```python
print("Columns:", df.columns.tolist())
print()
print(df.dtypes)
```

```python
print("Missing values per column:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
print("Fully duplicated rows (every column identical):", df.duplicated().sum())
dup_cols = ["Date", "Lecture Number", "Semester", "Branch", "Section"]
true_duplicates = df.duplicated(subset=dup_cols, keep=False).sum()
print("Rows that are true duplicates (same Date+Lecture+Semester+Branch+Section):", true_duplicates)
print("-> Since this is 0, every row represents one real, distinct lecture. Nothing needs to be dropped.")
```

```python
print(df.describe())
```

```python
# Verify the PDF's attendance formula: Attendance % = (Present / Enrolled) * 100
calculated_pct = (df["Students Present"] / df["Total Enrolled Students"] * 100).round(2)
mismatch_count = (calculated_pct - df["Attendance Percentage"]).abs().gt(0.5).sum()
print("Rows where the stored Attendance % does not match (Present/Enrolled)*100:", mismatch_count)
print("-> 0 mismatches means the Attendance Percentage column is already calculated correctly.")
```

## Step 4: Data Cleaning

**Faculty Experience** is the only column with missing values (167 rows). We fill
those with the median experience of the faculty we do have data for.


```python
median_experience = df["Faculty Experience"].median()
df["Faculty Experience"] = df["Faculty Experience"].fillna(median_experience)
print("Missing Faculty Experience values remaining:", df["Faculty Experience"].isnull().sum())
```

```python
# Parse the Date column properly (it is stored as DD-MM-YYYY text)
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True)

# Lecture duration is known in advance from the timetable, so it is a safe feature
start_t = pd.to_datetime(df["Start Time"].astype(str), format="mixed")
end_t = pd.to_datetime(df["End Time"].astype(str), format="mixed")
df["Lecture Duration (Minutes)"] = (end_t - start_t).dt.total_seconds() / 60

print(df[["Date", "Start Time", "End Time", "Lecture Duration (Minutes)"]].head())
```

## Step 5: Feature Engineering




```python
# Sort chronologically PER SECTION before computing any lag/rolling feature
df = df.sort_values(["Section", "Date", "Lecture Number"]).reset_index(drop=True)

# 5.1 Day of Semester / Week Number / Month
df["Day of Semester"] = (df["Date"] - df.groupby("Semester")["Date"].transform("min")).dt.days + 1
df["Week Number"] = df["Date"].dt.isocalendar().week.astype(int)
df["Month"] = df["Date"].dt.month

df[["Semester", "Date", "Day of Semester", "Week Number", "Month"]].drop_duplicates().head()
```

```python
# 5.2 Rolling average of the PREVIOUS 3 lectures' attendance, per Section
df["Previous Attendance Rolling Average"] = (
    df.groupby("Section")["Attendance Percentage"]
      .transform(lambda s: s.shift(1).rolling(window=3, min_periods=1).mean())
)

df[["Section", "Date", "Attendance Percentage", "Previous Lecture Attendance",
    "Previous Attendance Rolling Average"]].head(10)
```

```python
# 5.3 Consecutive Lecture Count
streak_break = (
    (df["Section"] != df["Section"].shift(1)) |
    (df["Date"] != df["Date"].shift(1)) |
    (df["Gap Since Previous Lecture"] > 5)
)
df["Consecutive Lecture Count"] = streak_break.cumsum()
df["Consecutive Lecture Count"] = df.groupby("Consecutive Lecture Count").cumcount() + 1

df[["Section", "Date", "Lecture Number", "Gap Since Previous Lecture",
    "Consecutive Lecture Count"]].head(10)
```

```python
# 5.4 Days Since Last Holiday
unique_dates = df[["Date"]].drop_duplicates().sort_values("Date").reset_index(drop=True)
holiday_flag_dates = set(df.loc[df["Holiday Before/After"] == "Yes", "Date"])
unique_dates["Is_Holiday_Adjacent"] = unique_dates["Date"].isin(holiday_flag_dates)

last_holiday = None
days_since_list = []
for _, row in unique_dates.iterrows():
    days_since_list.append(np.nan if last_holiday is None else (row["Date"] - last_holiday).days)
    if row["Is_Holiday_Adjacent"]:
        last_holiday = row["Date"]

unique_dates["Days Since Last Holiday"] = days_since_list
df = df.merge(unique_dates[["Date", "Days Since Last Holiday"]], on="Date", how="left")
df["Days Since Last Holiday"] = df["Days Since Last Holiday"].fillna(df["Days Since Last Holiday"].median())

df[["Date", "Holiday Before/After", "Days Since Last Holiday"]].drop_duplicates("Date").head(10)
```

```python
# 5.5 Monthly average attendance 
monthly_avg = df.groupby("Month")["Attendance Percentage"].mean()
print("Monthly average attendance:")
print(monthly_avg.round(2))
```

```python
# 5.6 Simple Yes/No -> 1/0 flags, and the classification target
df["Exam Week Flag"] = (df["Internal Test Week"] == "Yes").astype(int)
df["Assignment Due Flag"] = (df["Assignment Due"] == "Yes").astype(int)
df["Holiday Flag"] = (df["Holiday Before/After"] == "Yes").astype(int)
df["Special Event Flag"] = (df["Special Event"] == "Yes").astype(int)

def attendance_category(value):
    if value < 50:
        return "Low"
    elif value <= 75:
        return "Medium"
    return "High"

df["Attendance Category"] = df["Attendance Percentage"].apply(attendance_category)
print(df["Attendance Category"].value_counts())
```

```python
# Save the cleaned + feature-engineered dataset for submission
df_export = df.copy()
df_export["Date"] = df_export["Date"].dt.strftime("%d-%m-%Y")
df_export.to_csv("cleaned_attendance.csv", index=False)
print("Saved cleaned_attendance.csv, shape:", df_export.shape)
```

## Step 7: Model Preparation



```python
DROP_COLS = [
    "Attendance Percentage", "Students Present", "Attendance Category",
    "Date", "Start Time", "End Time", "Data Source", "Timetable Status",
]

feature_cols = [c for c in df.columns if c not in DROP_COLS]
categorical_cols = df[feature_cols].select_dtypes(include=["object"]).columns.tolist()
numeric_cols = [c for c in feature_cols if c not in categorical_cols]

print("Categorical columns:", categorical_cols)
print()
print("Numeric columns:", numeric_cols)
```

```python
preprocessor = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
    ]), numeric_cols),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore")),
    ]), categorical_cols),
])
print("Preprocessor ready.")
```

### Train/Test Split - Time-Based, Not Random



```python
df_sorted = df.sort_values(["Date", "Start Time"]).reset_index(drop=True)
split_idx = int(len(df_sorted) * 0.8)

X_all = df_sorted[feature_cols]
y_reg_all = df_sorted["Attendance Percentage"]
y_class_all = df_sorted["Attendance Category"]

X_train, X_test = X_all.iloc[:split_idx], X_all.iloc[split_idx:]
y_train, y_test = y_reg_all.iloc[:split_idx], y_reg_all.iloc[split_idx:]
y_train_class, y_test_class = y_class_all.iloc[:split_idx], y_class_all.iloc[split_idx:]

print("Training lectures:", X_train.shape[0], " | Testing lectures:", X_test.shape[0])
print("Train date range:", df_sorted["Date"].iloc[:split_idx].min(), "to", df_sorted["Date"].iloc[:split_idx].max())
print("Test date range :", df_sorted["Date"].iloc[split_idx:].min(), "to", df_sorted["Date"].iloc[split_idx:].max())
```

## Step 8: Model Training - Regression



```python
regression_models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=200, random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "SVM": SVR(kernel="rbf"),
    "XGBoost": XGBRegressor(n_estimators=200, random_state=42, objective="reg:squarederror"),
}

fitted_reg_models = {}
reg_predictions = {}

for name, estimator in regression_models.items():
    pipe = Pipeline([("preprocessor", preprocessor), ("model", estimator)])
    pipe.fit(X_train, y_train)
    reg_predictions[name] = pipe.predict(X_test)
    fitted_reg_models[name] = pipe
    print("Trained:", name)
```

## Step 9: Model Evaluation - Regression

Metrics required by the PDF: **MAE, RMSE, MAPE, R² Score.**


```python
reg_rows = []
for name, pred in reg_predictions.items():
    mae = mean_absolute_error(y_test, pred)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    mape = mean_absolute_percentage_error(y_test, pred) * 100
    r2 = r2_score(y_test, pred)
    reg_rows.append({"Model": name, "MAE": mae, "RMSE": rmse, "MAPE": mape, "R2": r2})

regression_results = pd.DataFrame(reg_rows).round(4)
regression_results
```

```python
plt.bar(regression_results["Model"], regression_results["MAE"])
plt.xticks(rotation=30)
plt.ylabel("MAE (lower is better)")
plt.title("Regression Model Comparison - MAE")
plt.tight_layout()
plt.show()
```

## Step 10: Model Comparison and Best Regression Model

```python
best_reg_row = regression_results.loc[regression_results["MAE"].idxmin()]
best_reg_name = best_reg_row["Model"]
print("Best regression model (lowest MAE):", best_reg_name)
regression_results.sort_values("MAE")
```

## Step 11: Classification Approach (Low / Medium / High)




```python
label_encoder = LabelEncoder()
y_train_enc = label_encoder.fit_transform(y_train_class)
y_test_enc = label_encoder.transform(y_test_class)
n_classes = len(label_encoder.classes_)

print("Classes present in this dataset:", list(label_encoder.classes_))
print("(If 'Low' does not appear, it is because no lecture in this dataset ever had attendance below 50%.)")
```

```python
classification_models = {
    "Logistic Regression": LogisticRegression(max_iter=2000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "SVM": SVC(probability=True, random_state=42),
    "k-NN": KNeighborsClassifier(n_neighbors=5),
    "Naive Bayes": GaussianNB(),
    "XGBoost": (
        XGBClassifier(n_estimators=200, random_state=42, eval_metric="mlogloss",
                      objective="multi:softprob", num_class=n_classes)
        if n_classes > 2 else
        XGBClassifier(n_estimators=200, random_state=42, eval_metric="logloss")
    ),
}

fitted_class_models = {}
class_rows = []

for name, estimator in classification_models.items():
    steps = [("preprocessor", preprocessor)]
    if name == "Naive Bayes":
        steps.append(("to_dense", FunctionTransformer(to_dense_array)))
    steps.append(("model", estimator))
    pipe = Pipeline(steps)

    pipe.fit(X_train, y_train_enc)
    pred = pipe.predict(X_test)
    proba = pipe.predict_proba(X_test)
    fitted_class_models[name] = pipe

    acc = accuracy_score(y_test_enc, pred)
    prec = precision_score(y_test_enc, pred, average="weighted", zero_division=0)
    rec = recall_score(y_test_enc, pred, average="weighted", zero_division=0)
    f1 = f1_score(y_test_enc, pred, average="weighted", zero_division=0)


    if n_classes == 2:
        roc_auc = roc_auc_score(y_test_enc, proba[:, 1])
    else:
        roc_auc = roc_auc_score(y_test_enc, proba, multi_class="ovr", average="weighted")

    class_rows.append({"Model": name, "Accuracy": acc, "Precision": prec,
                        "Recall": rec, "F1 Score": f1, "ROC-AUC": roc_auc})
    print("Trained:", name)

classification_results = pd.DataFrame(class_rows).round(4)
classification_results
```

## Step 12: Model Comparison and Best Classification Model


```python
best_class_row = classification_results.loc[classification_results["F1 Score"].idxmax()]
best_class_name = best_class_row["Model"]
print("Best classification model (highest F1 Score):", best_class_name)

plt.bar(classification_results["Model"], classification_results["Accuracy"])
plt.xticks(rotation=30)
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.title("Classification Model Comparison - Accuracy")
plt.tight_layout()
plt.show()
```

### Confusion Matrix for the Best Classification Model


```python
best_pipe = fitted_class_models[best_class_name]
pred_best = best_pipe.predict(X_test)
pred_best_labels = label_encoder.inverse_transform(pred_best)

cm = confusion_matrix(y_test_class, pred_best_labels, labels=label_encoder.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=label_encoder.classes_)
disp.plot()
plt.title(f"Confusion Matrix - {best_class_name}")
plt.tight_layout()
plt.show()
```

## Step 13: Final Prediction and Saving Models for Streamlit


```python
print("Example predictions on the first 5 test lectures:")
example = pd.DataFrame({"Actual Attendance %": y_test.iloc[:5].values})
for name, pred in reg_predictions.items():
    example[name] = pred[:5].round(2)
example
```

```python
# Save every trained model separately, as its own .pkl file, so each
# algorithm required by the PDF (Phase 3) can be loaded and used
# independently in the Streamlit app - not just the single best one.
reg_filename_map = {
    "Linear Regression": "linear_model.pkl",
    "Decision Tree": "decision_tree_model.pkl",
    "Random Forest": "random_forest_model.pkl",
    "Gradient Boosting": "gradient_model.pkl",
    "SVM": "svm_model.pkl",
    "XGBoost": "xgb_reg_model.pkl",
}
for name, filename in reg_filename_map.items():
    joblib.dump(fitted_reg_models[name], filename)

class_filename_map = {
    "Logistic Regression": "logistic_model.pkl",
    "Decision Tree": "decision_tree_classifier_model.pkl",
    "Random Forest": "random_forest_classifier_model.pkl",
    "SVM": "svm_classifier_model.pkl",
    "k-NN": "knn_classifier_model.pkl",
    "Naive Bayes": "naive_bayes_model.pkl",
    "XGBoost": "xgb_classifier_model.pkl",
}
for name, filename in class_filename_map.items():
    joblib.dump(fitted_class_models[name], filename)

joblib.dump(label_encoder, "label_encoder.pkl")

regression_results.to_csv("regression_results.csv", index=False)
classification_results.to_csv("classification_results.csv", index=False)

print("Saved one .pkl file per trained algorithm (regression + classification),")
print("plus the label encoder and both results tables.")
print("Best regression model (lowest MAE):", best_reg_name)
print("Best classification model (highest F1 Score):", best_class_name)
```
